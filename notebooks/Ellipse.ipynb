```python
#Các hàm cần thiết để tính toán
import numpy as np
from numpy import pi
import matplotlib.pyplot as plt
import csv 
import math
```

```python
e = math.e

def calE(x, y, z, lc, dc, delta, xx, zz):
    # yy = l / 2 * ((1 - (zz / d) ^ 2) ^ 0.5)
    # calE = Exp(zz / sicma) * (z - zz) * (1 / ((x - xx) ^ 2 + (y + yy) ^ 2 + (z - zz) ^ 2) ^ 1.5 - 1 / ((x - xx) ^ 2 + (y - yy) ^ 2 + (z - zz) ^ 2) ^ 1.5)
    yy = lc / 2 * ((1 - (zz / dc) ** 2) ** 0.5)
    out = e ** (zz / delta) * (z - zz) * (1 / ((x - xx) ** 2 + (y + yy) ** 2 + (z - zz) ** 2) ** 1.5 - 1 / (
                (x - xx) ** 2 + (y - yy) ** 2 + (z - zz) ** 2) ** 1.5)
    return out


def calE_add1(x, y, z, wc, lc, dc, delta, yy, zz):
    # tangalfa = yy * (2 * d / l) ^ 2 / (zz + 0.000000001)
    # calE_add1 = tangalfa / ((1 + tangalfa ^ 2) ^ 0.5) * Exp(zz / sicma) * (z - zz) * (-2 * yy / l) * (1 / ((x + w / 2) ^ 2 + (y - yy) ^ 2 + (z - zz) ^ 2) ^ 1.5 + 1 / ((x - w / 2) ^ 2 + (y - yy) ^ 2 + (z - zz) ^ 2) ^ 1.5)
    tangalfa = yy * (2 * dc / lc) ** 2 / (zz + 0.000000001)
    out = tangalfa / ((1 + tangalfa ** 2) ** 0.5) * e ** (zz / delta) * (z - zz) * (-2 * yy / lc) * (
                1 / ((x + wc / 2) ** 2 + (y - yy) ** 2 + (z - zz) ** 2) ** 1.5 + 1 / (
                    (x - wc / 2) ** 2 + (y - yy) ** 2 + (z - zz) ** 2) ** 1.5)
    return out


def calE_add2(x, y, z, wc, lc, dc, delta, yy, zz):
    # tangalfa = -yy * (2 * d / l) ^ 2 / (zz + 0.000000001)
    # calE_add2 = tangalfa / ((1 + tangalfa ^ 2) ^ 0.5) * Exp(zz / sicma) * (z - zz) * (2 * yy / l) * (1 / ((x + w / 2) ^ 2 + (y - yy) ^ 2 + (z - zz) ^ 2) ^ 1.5 + 1 / ((x - w / 2) ^ 2 + (y - yy) ^ 2 + (z - zz) ^ 2) ^ 1.5)
    tangalfa = -yy * (2 * dc / lc) ** 2 / (zz + 0.000000001)
    out = tangalfa / ((1 + tangalfa ** 2) ** 0.5) * e ** (zz / delta) * (z - zz) * (2 * yy / lc) * (
                1 / ((x + wc / 2) ** 2 + (y - yy) ** 2 + (z - zz) ** 2) ** 1.5 + 1 / (
                    (x - wc / 2) ** 2 + (y - yy) ** 2 + (z - zz) ** 2) ** 1.5)
    return out


def transform(x, y, x_offset, y_offset, angle):
    xr = (x - x_offset) * math.cos(angle * math.pi / 180) - (y - y_offset) * math.sin(angle * math.pi / 180)
    yr = (x - x_offset) * math.sin(angle * math.pi / 180) + (y - y_offset) * math.cos(angle * math.pi / 180)
    return xr, yr


def InteE(x, y, z, wc, lc, dc, delta):
    n = 20
    m = 20
    a = -wc / 2
    b = wc / 2
    c = -dc
    d = 0

    h = (d - c) / n
    k = (b - a) / m

    T2 = 0
    T3 = 0
    T4 = 0

    T1 = calE(x, y, z, lc, dc, delta, a, c) + calE(x, y, z, lc, dc, delta, b, c) + calE(x, y, z, lc, dc, delta, a,
                                                                                        d) + calE(x, y, z, lc, dc,
                                                                                                  delta, b, d)

    for j in range(1, n):
        yj = c + j * h
        T3 = T3 + calE(x, y, z, lc, dc, delta, a, yj) + calE(x, y, z, lc, dc, delta, b, yj)

    for i in range(1, m):
        xi = a + i * k
        T2 = T2 + calE(x, y, z, lc, dc, delta, xi, c) + calE(x, y, z, lc, dc, delta, xi, d)
        for j in range(1, n):
            yj = c + j * h
            T4 = T4 + calE(x, y, z, lc, dc, delta, xi, yj)

    out = h * k * (T1 / 4 + T2 / 2 + T3 / 2 + T4)

    return out


def InteE_add1(x, y, z, wc, lc, dc, delta):
    n = 20
    m = 20

    x_ = np.zeros(m + 1)
    y_ = np.zeros((n + 1, m + 1))
    c = np.zeros(m + 1)
    d = np.zeros(m + 1)
    h = np.zeros(m + 1)

    a = -lc / 2
    b = 0

    k = (b - a) / m

    T2 = 0
    T3 = 0
    T4 = 0

    for j in range(m + 1):
        x_[j] = a + j * k
        c[j] = -dc * (1 - (2 * x_[j] / lc) ** 2) ** 0.5
        d[j] = 0
        h[j] = (d[j] - c[j]) / n
        for i in range(n + 1):
            y_[i, j] = c[j] + i * h[j]

    T1 = h[0] * (calE_add1(x, y, z, wc, lc, dc, delta, x_[0], c[0]) + calE_add1(x, y, z, wc, lc, dc, delta, x_[0],
                                                                                d[0])) + h[m] * (
                     calE_add1(x, y, z, wc, lc, dc, delta, x_[m], c[m]) + calE_add1(x, y, z, wc, lc, dc, delta, x_[m],
                                                                                    d[m]))

    for i in range(n):
        T3 = T3 + h[0] * calE_add1(x, y, z, wc, lc, dc, delta, x_[0], y_[i, 0]) + h[m] * calE_add1(x, y, z, wc, lc, dc,
                                                                                                   delta, x_[m],
                                                                                                   y_[i, m])

    for j in range(m):

        T2 = T2 + (calE_add1(x, y, z, wc, lc, dc, delta, x_[j], c[j]) + calE_add1(x, y, z, wc, lc, dc, delta, x_[j],
                                                                                  d[j])) * h[j]

        for i in range(n):
            T4 = T4 + (calE_add1(x, y, z, wc, lc, dc, delta, x_[j], y_[i, j])) * h[j]

    T1 = k * T1 / 4
    T2 = k * T2 / 2
    T3 = k * T3 / 2
    T4 = k * T4

    out = (T1 + T2 + T3 + T4)
    return out


def InteE_add2(x, y, z, wc, lc, dc, delta):
    n = 20
    m = 20

    x_ = np.zeros(m + 1)
    y_ = np.zeros((n + 1, m + 1))
    c = np.zeros(m + 1)
    d = np.zeros(m + 1)
    h = np.zeros(m + 1)

    a = 0
    b = lc / 2

    k = (b - a) / m

    T2 = 0
    T3 = 0
    T4 = 0

    for j in range(m + 1):
        x_[j] = a + j * k
        c[j] = -dc * (1 - (2 * x_[j] / lc) ** 2) ** 0.5
        d[j] = 0
        h[j] = (d[j] - c[j]) / n
        for i in range(n + 1):
            y_[i, j] = c[j] + i * h[j]

    T1 = h[0] * (calE_add2(x, y, z, wc, lc, dc, delta, x_[0], c[0]) + calE_add2(x, y, z, wc, lc, dc, delta, x_[0],
                                                                                d[0])) + h[m] * (
                     calE_add2(x, y, z, wc, lc, dc, delta, x_[m], c[m]) + calE_add2(x, y, z, wc, lc, dc, delta, x_[m],
                                                                                    d[m]))

    for i in range(n):
        T3 = T3 + h[0] * calE_add2(x, y, z, wc, lc, dc, delta, x_[0], y_[i, 0]) + h[m] * calE_add2(x, y, z, wc, lc, dc,
                                                                                                   delta, x_[m],
                                                                                                   y_[i, m])

    for j in range(m):

        T2 = T2 + (calE_add1(x, y, z, wc, lc, dc, delta, x_[j], c[j]) + calE_add1(x, y, z, wc, lc, dc, delta, x_[j],
                                                                                  d[j])) * h[j]

        for i in range(n):
            T4 = T4 + (calE_add1(x, y, z, wc, lc, dc, delta, x_[j], y_[i, j])) * h[j]

    T1 = k * T1 / 4
    T2 = k * T2 / 2
    T3 = k * T3 / 2
    T4 = k * T4

    out = (T1 + T2 + T3 + T4)
    return out
```

```python
wc_all = np.linspace(0.3,1.5,12)
lc_all = np.linspace(2,20,19)
dc_all = np.linspace(0.1,3,29)
```

```python
N = 16
Res = 0.78  # Resolution mm
aR = np.linspace(-N * Res, N * Res, 2 * N )
bR = np.linspace(-N * Res, N * Res, 2 * N )
H = np.zeros((len(aR), len(bR)))
x_offset=0
y_offset=0
angle= 0
wc=0.9
lc=10
dc=3
z=1
freq=5000
sicma=35461000
mu=0.0000012566
csi=0.085
K=1.5
I=0.01
G=3981

delta = 1 / np.sqrt(pi * freq * mu * sicma) * 1000

for wc in wc_all:
  for lc in lc_all:
    for dc in dc_all:
      tam = 'Ellipse' + '_' + str(wc) + '_' + str(lc) + '_' + str(dc) + '.csv'
      if(tam not in  os.listdir(os.path.expanduser(path1))):
        print(tam)
        H = np.zeros((len(aR), len(bR)))
        for u, xo in enumerate(aR):
          for v, yo in enumerate(aR):
            x, y = transform(xo, yo, x_offset, y_offset, angle)
            H[u, v] = InteE(x, y, z, wc, lc, dc, delta) + InteE_add1(x, y, z, wc, lc, dc, delta) - InteE_add2(x, y, z, wc, lc, dc, delta)
            H[u, v] = math.cos(angle * math.pi / 180) * H[u, v]

        H = csi/4/pi*H
        H = K*I*G*H
        name_params=['N','liff_off','Csi','K','I','f','sicma','mu','G','Res']
        params=[32, z, csi, K, I, freq, sicma, mu, G, Res]
        L = ['shape', 'width', 'length', 'depth','angle','x_offset','y_offset']
        C = ['Ellipse', wc, lc, dc, angle, x_offset, y_offset]

        filename = '/content/gdrive/MyDrive/Dipole Model/Data/Low Data/' + C[0] + '_' + str(C[1]) + '_' + str(C[2]) + '_' + str(C[3]) + '.csv'

        with open(filename, 'w', newline='') as f:
          writer = csv.writer(f)
          writer.writerow(name_params)
          writer.writerow(params)
          writer.writerow(L)
          writer.writerow(C)
          writer.writerows(H)
          f.close()
```
